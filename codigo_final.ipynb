```python
import random


class Nivel:
    def __init__(self, nombre, palabras, intentos_maximos):
        self.nombre = nombre
        self.palabras = palabras
        self.intentos_maximos = intentos_maximos

    def obtener_palabra_aleatoria(self):
        return random.choice(self.palabras)

    def obtener_intentos(self):
        return self.intentos_maximos


class NivelFacil(Nivel):
    def __init__(self):
        super().__init__("FÁCIL", ["CASA", "GATO", "PERRO", "MESA", "FLOR"], 8)


class NivelNormal(Nivel):
    def __init__(self):
        super().__init__("NORMAL", ["ESCUELA", "CUADERNO", "VENTANA", "FAMILIA", "AMIGOS"], 6)


class NivelDificil(Nivel):
    def __init__(self):
        super().__init__("DIFÍCIL", ["COMPUTADOR", "MARIPOSA", "BIBLIOTECA", "PROGRAMACION", "ELEFANTE"], 5)


class NivelExperto(Nivel):
    def __init__(self):
        super().__init__("EXPERTO", ["DESARROLLO", "INTELIGENCIA", "UNIVERSIDAD", "TECNOLOGIA", "APRENDIZAJE"], 4)


class InterfazConsola:
    def __init__(self):
        self.dibujos = [
            """
             -----
             |   |
             |
             |
             |
            ---
            """,
            """
             -----
             |   |
             |   O
             |
             |
            ---
            """,
            """
             -----
             |   |
             |   O
             |   |
             |
            ---
            """,
            """
             -----
             |   |
             |   O
             |  /|
             |
            ---
            """,
            """
             -----
             |   |
             |   O
             |  /|\\
             |
            ---
            """,
            """
             -----
             |   |
             |   O
             |  /|\\
             |  /
            ---
            """,
            """
             -----
             |   |
             |   O
             |  /|\\
             |  / \\
            ---
            """
        ]

    def mostrar_bienvenida(self):
        print("==============================")
        print("      JUEGO DEL AHORCADO")
        print("==============================")

    def seleccionar_nivel(self):
        print("\nSelecciona el nivel que deseas jugar:")
        print("1. Fácil")
        print("2. Normal")
        print("3. Difícil")
        print("4. Experto")
        return input("Escribe el número del nivel: ")

    def pedir_letra(self):
        return input("Ingresa una letra: ").upper()

    def mostrar_dibujo(self, errores, intentos):
        posicion = errores * (len(self.dibujos) - 1) // intentos
        print(self.dibujos[posicion])

    def mostrar_resultado(self, gano, palabra):
        if gano:
            print("\n¡FELICIDADES! GANASTE.")
            print("La palabra era:", palabra)
        else:
            print("\n¡PERDISTE!")
            print("La palabra correcta era:", palabra)


class JuegoAhorcado:
    def __init__(self):
        self.interfaz = InterfazConsola()
        self.nivel_actual = None
        self.palabra_secreta = ""
        self.letras_adivinadas = []
        self.letras_usadas = []
        self.errores = 0

    def configurar_nivel(self, nivel_id):
        niveles = {
            "1": NivelFacil(),
            "2": NivelNormal(),
            "3": NivelDificil(),
            "4": NivelExperto()
        }
        return niveles.get(nivel_id)

    def mostrar_pistas(self):
        print("\n¡Comienza el juego!")
        print("Pista: La palabra comienza con la letra:", self.palabra_secreta[0])
        print("La palabra tiene", len(self.palabra_secreta), "letras.")

    def mostrar_estado(self):
        palabra_mostrada = ""
        for letra in self.palabra_secreta:
            if letra in self.letras_adivinadas:
                palabra_mostrada = palabra_mostrada + letra + " "
            else:
                palabra_mostrada = palabra_mostrada + "_ "

        print("\nPalabra:", palabra_mostrada)
        print("Letras utilizadas:", self.letras_usadas)
        print("Intentos restantes:", self.nivel_actual.obtener_intentos() - self.errores)

    def verificar_victoria(self):
        return all(letra in self.letras_adivinadas for letra in self.palabra_secreta)

    def procesar_letra(self, letra):
        if len(letra) != 1 or not letra.isalpha():
            print("Ingresa solamente una letra.")
            return

        if letra in self.letras_usadas:
            print("Ya utilizaste esa letra.")
            return

        self.letras_usadas.append(letra)

        if letra in self.palabra_secreta:
            print("¡Correcto! La letra está en la palabra.")
            self.letras_adivinadas.append(letra)
        else:
            print("Incorrecto. Pierdes un intento.")
            self.errores = self.errores + 1

        self.interfaz.mostrar_dibujo(self.errores, self.nivel_actual.obtener_intentos())

    def iniciar_partida(self):
        self.interfaz.mostrar_bienvenida()
        jugar = "SI"

        while jugar == "SI":
            opcion = self.interfaz.seleccionar_nivel()
            self.nivel_actual = self.configurar_nivel(opcion)

            if self.nivel_actual is None:
                print("Nivel no válido.")
                continue

            print("Elegiste el nivel", self.nivel_actual.nombre)
            self.palabra_secreta = self.nivel_actual.obtener_palabra_aleatoria()
            self.letras_adivinadas = []
            self.letras_usadas = []
            self.errores = 0

            self.mostrar_pistas()
            intentos = self.nivel_actual.obtener_intentos()

            while self.errores < intentos:
                self.mostrar_estado()

                if self.verificar_victoria():
                    break

                letra = self.interfaz.pedir_letra()
                self.procesar_letra(letra)

            self.interfaz.mostrar_resultado(self.verificar_victoria(), self.palabra_secreta)
            jugar = input("\n¿Deseas jugar otra vez? SI / NO: ").upper()

        print("\nGracias por jugar al AHORCADO.")


juego = JuegoAhorcado()
juego.iniciar_partida()
```
